```python
#An online platform tracks time spent (in minutes) by students on 6 courses over 14 days using a NumPy array.
# Task:Compute total time spent per course,
# Identify courses where average daily engagement exceeds a threshold,
# Rank courses based on engagement



import numpy as np

time_spent=np.array([[120,80,70,30,60,40],[130,40,60,80,40,60],[30,60,75,60,70,30],[120,140,50,90,40,45],[80,90,50,30,40,80],[130,40,60,76,56,65],[68,85,64,65,30,74],[50,30,50,60,30,90],[40,80,45,80,30,90],
            [30,80,30,80,30,73],[70,50,30,50,20,89],[90,45,53,34,40,67],[130,40,60,80,40,78],[130,40,60,80,40,45]])



total_time_per_course=np.sum(time_spent,axis=0)

print(f"Total time per course 1: {total_time_per_course[0]}")
print(f"Total time per course 2: {total_time_per_course[1]}")
print(f"Total time per course 3: {total_time_per_course[2]}")
print(f"Total time per course 4: {total_time_per_course[3]}")
print(f"Total time per course 5: {total_time_per_course[4]}")
print(f"Total time per course 6: {total_time_per_course[5]}")

threshold=80

avg_time=list(np.average(time_spent,axis=0))
print("\nAverage time",avg_time)

for a in avg_time:
    if a> threshold:
        print(f"\nCourse {avg_time.index(a)+1} exceeds threshold {threshold}\n")

sorted_avg=list(avg_time)
sorted_avg.sort(reverse=True)
print(sorted_avg)
for i in sorted_avg:
    print(f"Rank {sorted_avg.index(i)+1}: Course {avg_time.index(i)+1} ")

    
```
