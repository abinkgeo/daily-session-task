```python
'''Online Course Completion AnalysisScenario
A learning platform wants to understand course engagement and completion behavior.
Problem Statement
Create a dataset of at least 300 course enrollments with:

student_id, course_name, progress_percent, completion_days, rating

Using Pandas:

Identify courses with poor completion performance.

Compute average completion time per course.

Using NumPy:

Perform statistical analysis on completion duration.

Simulate improved progress scenarios.

Using a visualization library:

Visualize completion time distribution.

Compare average ratings across courses.'''
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 

np.random.seed(1)
num=300
course_choice=["Java","Web Development","Python","Data Science"]

data={
    "student_id":range(1,301),
    "course_name":np.random.choice(course_choice,num),
    "progress_percent":np.random.randint(30,101,num),
    "completion_date":np.random.randint(30,121,num),
    "rating":np.random.randint(1,6,num)
}

df=pd.DataFrame(data)

df.head()

```

```python
# poor completion

poor_completion = df.groupby("course_name")["completion_date"].mean()

poor_completion[poor_completion<75]



```

```python
# average completion time per course

avg_completion=df.groupby("course_name")["progress_percent"].mean()

avg_completion
```

```python
# statistical analysis on completion duration

arr=np.array(df["completion_date"])
print("Mean of completion",np.mean(arr))
print("Median of completion",np.median(arr))
print("Standard deviation",np.std(arr))
print("Min value",np.min(arr))
print("Max value",np.max(arr))
print("Count",np.count_nonzero(arr))


# print(df["completion_date"].describe())
```

```python
# Simulate improved progress scenarios.

progress=np.array((df["progress_percent"]*1.15).clip(0,100))

df["improve_progress"]=progress

print(df["progress_percent"].describe())
print()
print(df["improve_progress"].describe())



```

```python
# Visualize completion time distribution.


plt.hist(df["completion_date"],bins=10,color="yellow",edgecolor="red")
plt.xlabel("Completion Days")
plt.ylabel("No of Students")
plt.title("Completion Time")
plt.xlim(30,120)
plt.show()
```

```python
# Avg rating accross Course

avg_rating=df.groupby("course_name")["rating"].mean()

colors=["orange","gray","blue","green"]
avg_rating.plot(kind="bar",xlabel="Course Name",ylabel="Avg ratings",title="Average rating across Course",color=colors)

plt.ylim(0,6)
plt.show()

```
