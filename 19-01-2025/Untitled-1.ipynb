```python
#Given a DataFrame with some NaN values in age:

# Replace missing ages with the mean age


import pandas as pd
import numpy as np

num=50
np.random.seed(1)
data={
    "age":np.random.randint(1,50,num)

}

df=pd.DataFrame(data)
mask=np.random.rand(*df.shape) < 0.2 

df[mask]=np.nan

mean_age=df["age"].mean()


df["age"]=df["age"].fillna(mean_age)

df




```
